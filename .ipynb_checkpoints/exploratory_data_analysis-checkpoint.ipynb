```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

First, we will load in the data we saved from the preprocessing section:
- full_df will be the full clean dataframe
- surface_df will be the surface temperatures only, with one row per ARGO profile
- sst_df will be the raw satellite temp data, for use while plotting

```python
# load in the preprocessed data that we saved
full_df = pd.read_csv('clean_data_gulfstream_jan_2023.csv')
full_df['TIME'] = pd.to_datetime(full_df['TIME'])

surface_df = pd.read_csv('clean_SURFACE_data_gulfstream_jan_2023.csv')
surface_df['TIME'] = pd.to_datetime(surface_df['TIME'])

# as well as the full SST dataframe
sst_df = pd.read_csv('SST_only_gulfstream_jan_2023_not_clean.csv')

sst_df.head()
```

We will look at some summary statistics for the full dataframe as well as the surface dataframe

```python
# .info() gives data type and non-null count for each column, and memory usage for the dataframe
# .describe() gives some summary statistics to explore the data
print('\nsummary statistics for numeric columns of full dataframe:\n')
print(full_df.describe())
```

```python
# .describe() gives some summary statistics to explore the data
print('\nsummary statistics for numeric columns for surface dataframe:\n')
print(surface_df.describe())
```

Some observations:
- in full_df, pressure goes from 0 to 1000 dbar which is roughly 1000 m deep (1 dbar ~ 1 meter depth in ocean)
- the DIFF column in surface_df (which is ARGO measurement - satellite measurement) goes from -7 to 4°C, mean and median are close to 0°C. This indicates that the difference may be random, because error is centered at zero

Next, we can get into some plotting. This first plot will plot the difference between the ARGO temp measurement and the satellite temp measurement for each ARGO float location, and I will plot that on top of the satellite SST data to visualize where the gulf stream is

```python
# plot: (Argo SST - satellite SST), overlain on satellite SST

fig, ax = plt.subplots(1,1,figsize=(8,6), layout='constrained')

# plot satellite SST
sst = ax.scatter(sst_df['longitude'], sst_df['latitude'], c=sst_df['sst'], cmap='plasma')
fig.colorbar(sst, ax=ax, label='Satellite SST (°C)', shrink=0.6)

# plot differences between temps
diffs = ax.scatter(surface_df['LON'], surface_df['LAT'], c=surface_df['DIFF'], cmap='Greys',s=70,edgecolors='black')
fig.colorbar(diffs, ax=ax, label='Diff btwn ARGO Satellite SST (°C)', shrink=0.6)

ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.set_title('Difference between ARGO and Satellite SST Measurements\nPlotted over Satellite SST Measurements')

ax.set_aspect('equal', adjustable='box')
#ax.grid()
```

It looks like the highest differences in agreement are around where the gulf stream is, where there is a pretty abrupt change in SST. My initial hypothesis is that the satellite resolution does not have the resolution to capture smaller-scale eddies coming off the gulf stream

Next, I will plot some of the individual ARGO profiles, at four points. I will find the profiles closest to visually identified coordinates, with two near the gulf stream and two farther from the gulf stream

```python
# one row per profile (float + cycle), with its location and TEMP_MINUS_SST
profiles = (full_df.groupby(['PLATFORM_NUMBER', 'CYCLE_NUMBER'])
            .agg(LAT=('LATITUDE', 'mean'), LON=('LONGITUDE', 'mean'),
                 TEMP_MINUS_SST=('TEMP_MINUS_SST', 'mean'))
            .reset_index())

targets = [(37.5, -71.5), (40.0, -68.0), (30.0, -67.0), (28.0, -52.0)]

fig, axes = plt.subplots(2, 2, figsize=(10, 8), sharex=True, sharey=True, layout='constrained')

true_loc_list = []
loc_name = 0

for ax, (lat, lon) in zip(axes.flat, targets):
    # update loc_name
    loc_name +=1
    # closest profile to the target (longitude scaled by cos(lat) so distances are fair)
    dist = np.hypot(profiles['LAT'] - lat, (profiles['LON'] - lon) * np.cos(np.deg2rad(lat)))
    p = profiles.loc[dist.idxmin()]

    prof = full_df[(full_df['PLATFORM_NUMBER'] == p['PLATFORM_NUMBER']) &
                   (full_df['CYCLE_NUMBER'] == p['CYCLE_NUMBER'])]

    ax.scatter(prof['TEMP'], prof['PRES'], s=12, c='tab:red', alpha=0.7)
    ax.set_title(f"Location {loc_name}\nFloat #: {int(p['PLATFORM_NUMBER'])}, cycle {int(p['CYCLE_NUMBER'])}: "
                 f"({p['LAT']:.2f}, {p['LON']:.2f})\nARGO−Sat Temp = {p['TEMP_MINUS_SST']:.2f} °C", fontsize=10)
    ax.grid(alpha=0.3)
    true_loc_list.append([p['LAT'],p['LON']])

axes[0, 0].invert_yaxis()   # axes are shared, so this flips all four (pressure increases downward)
for ax in axes[0, :]:
    ax.set_xlabel('In-situ Temperature (°C)', style='italic')
for ax in axes[1, :]:
    ax.set_xlabel('In-situ Temperature (°C)', style='italic')
for ax in axes[:, 0]:
    ax.set_ylabel('Pressure (dbar)', style='italic')

fig.suptitle('Temperature vs. Pressure for Selected ARGO Profiles')
plt.show()




### and plot where they are on the map
fig, ax = plt.subplots(1,1,figsize=(8,6))

# plot satellite SST
sst = ax.scatter(sst_df['longitude'], sst_df['latitude'], c=sst_df['sst'], cmap='plasma')
fig.colorbar(sst, ax=ax, label='Satellite SST (°C)', shrink=0.6)

# plot the float locations
loc_name=0
for latlon in true_loc_list:
    loc_name+=1 #update loc name
    lat = latlon[0]
    lon = latlon[1]
    plt.scatter(lon, lat, s=80, color='black', marker='*')
    plt.annotate(f'Loc {loc_name}', xy=(lon, lat))
    
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.set_title('Profile Locations')
ax.set_aspect('equal', adjustable='box')
```

I will plot the ARGO SST observations compared to the satellite SST observations next

```python
fig, ax = plt.subplots(1,1,figsize=(8,6))

ax.scatter(surface_df['SST'], surface_df['TEMP_SURF'])
x_axis = np.arange(np.min(surface_df['SST']), np.max(surface_df['SST']), 0.5)
ax.plot(x_axis, x_axis, linestyle='--', label='x=y', c='grey')
ax.legend()
ax.set_xlabel('Satellite SST (°C)')
ax.set_ylabel('ARGO SST (°C)')
ax.set_title('Satellite vs ARGO SST Measurements')
plt.show()
```

The two surface temperature measurements do look pretty clearly correlated, It's curious that they seem less aligned at cooler temperatures

Now I'll look at how the satellite SST compares to ARGO measurements at different depths. I will look at 50, 100, 500, and 1000 meters depth (assuming 1 dbar pressure is 1 m depth, and with a wiggle room of ±10m because each float took measurements at different exact depths)

```python
depths = [50, 100, 500, 1000] # meters (~ dbar)
tol = 10 # max allowed gap (dbar) between target depth and nearest measured level
profile_keys = ['PLATFORM_NUMBER', 'CYCLE_NUMBER']

fig, axes = plt.subplots(2, 2, figsize=(10, 8), sharex=True, layout='constrained')

for ax, depth in zip(axes.flat, depths):
    # for each profile, find the measurement level closest to the target depth
    tmp = full_df.assign(DZ=(full_df['PRES'] - depth).abs())
    nearest = tmp.loc[tmp.groupby(profile_keys)['DZ'].idxmin()]
    nearest = nearest[nearest['DZ'] <= tol]      # drop profiles that don't reach this depth

    ax.scatter(nearest['matched_sst'], nearest['TEMP'], s=20, alpha=0.7)

    # x=y reference line spanning the data in this panel
    lo = min(nearest['matched_sst'].min(), nearest['TEMP'].min())
    hi = max(nearest['matched_sst'].max(), nearest['TEMP'].max())
    ax.plot([lo, hi], [lo, hi], linestyle='--', c='grey', label='x=y')

    ax.set_title(f'ARGO at ~{depth} m (n = {len(nearest)})')
    ax.set_ylabel(f'ARGO Temp at ~{depth} m (°C)')
    ax.grid(alpha=0.3)
    ax.legend()

for ax in axes[1, :]:
    ax.set_xlabel('Satellite SST (°C)')

fig.suptitle('ARGO Temperature at Depth vs. Satellite SST')
plt.show()
```

At 50 and 100 meters deep, the ARGO temperatures look roughly linearly correlated with the satellite SST from about 10-25°C SST. This doesn't really hold at the lower temperatures though, and it also doesn't hold at the deeper depths, although there does seem to be a pattern still at those depths that we can characterize more when we are modeling.

#### General Overview of ARGO Float Data

Before investigating the differences between ARGO and satellite temperature measurements, we will examine the overall distribution and characteristics of the ARGO dataset. This includes the geographic distribution of float profiles, the distribution of measured temperatures and pressures, and the number of profiles collected by individual floats. These visualizations provide context for the more detailed analysis that follows.

```python
# General overview of ARGO float data

# make dataframe w/ one row per float profile
# each profile is ID by float and cycle #
profiles_overview = (
    full_df.groupby(['PLATFORM_NUMBER', 'CYCLE_NUMBER'])
    .agg(
        LATITUDE=('LATITUDE', 'mean'),
        LONGITUDE=('LONGITUDE', 'mean'),
        MEAN_TEMP=('TEMP', 'mean'),
        MEAN_PRESSURE=('PRES', 'mean')
    )
    .reset_index()
)

# basic info about dataset
print("ARGO FLOAT DATA OVERVIEW")
print("-" * 35)
print(f"Total measurements: {len(full_df):,}")
print(f"Unique floats: {full_df['PLATFORM_NUMBER'].nunique():,}")
print(f"Unique float profiles: {len(profiles_overview):,}")
print(f"Temperature range: {full_df['TEMP'].min():.2f} to "
      f"{full_df['TEMP'].max():.2f} °C")
print(f"Pressure range: {full_df['PRES'].min():.2f} to "
      f"{full_df['PRES'].max():.2f} dbar")

# create 2x2 figure
fig, axes = plt.subplots(2, 2, figsize=(12, 9),
                         layout='constrained')

# plot 1: geographic distribution of ARGO profiles
axes[0, 0].scatter(
    profiles_overview['LONGITUDE'],
    profiles_overview['LATITUDE'],
    s=12,
    alpha=0.6
)

axes[0, 0].set_xlabel('Longitude')
axes[0, 0].set_ylabel('Latitude')
axes[0, 0].set_title('Geographic Distribution of ARGO Profiles')
axes[0, 0].grid(alpha=0.3)

# plot 2: distribution of measured temperatures
axes[0, 1].hist(
    full_df['TEMP'].dropna(),
    bins=30,
    edgecolor='black'
)

axes[0, 1].set_xlabel('Temperature (°C)')
axes[0, 1].set_ylabel('Number of Measurements')
axes[0, 1].set_title('Distribution of ARGO Temperatures')
axes[0, 1].grid(axis='y', alpha=0.3)

# plot 3: distribution of measured pressures
axes[1, 0].hist(
    full_df['PRES'].dropna(),
    bins=30,
    edgecolor='black'
)

axes[1, 0].set_xlabel('Pressure (dbar)')
axes[1, 0].set_ylabel('Number of Measurements')
axes[1, 0].set_title('Distribution of ARGO Pressures')
axes[1, 0].grid(axis='y', alpha=0.3)

# plot 4: # of profiles collected by each float
profiles_per_float = (
    profiles_overview.groupby('PLATFORM_NUMBER')['CYCLE_NUMBER']
    .nunique()
    .sort_values(ascending=False)
    .head(15)
    .sort_values()
)

axes[1, 1].barh(
    profiles_per_float.index.astype(str),
    profiles_per_float.values
)

axes[1, 1].set_xlabel('Number of Profiles')
axes[1, 1].set_ylabel('Float Number')
axes[1, 1].set_title('Top 15 Floats by Number of Profiles')
axes[1, 1].grid(axis='x', alpha=0.3)

fig.suptitle('Overview of ARGO Float Data', fontsize=15)

plt.show()
```
